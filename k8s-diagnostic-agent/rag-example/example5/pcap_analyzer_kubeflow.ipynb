````python
import kfp
from kfp import dsl
from kfp import kubernetes
import os
````

## 1. Define Pipeline Components

````python
@dsl.component(
    base_image='amitinfo2k/pcap-analyzer:latest',    
    #packages_to_install=['pandas', 'scikit-learn', 'scapy', 'sentence-transformers']
)
def prepare_data(
    input_pcaps: str,
    mapping_file: str,
    output_dir: str
) -> str:
    import os
    from pcap_analyzer.modules.data_prep.pcap_processor import PCAPProcessor
    from pcap_analyzer.config import load_config
    
    os.makedirs(output_dir, exist_ok=True)
    features_file = os.path.join(output_dir, 'features.json')
    
    config = load_config()
    processor = PCAPProcessor(config, mapping_file=mapping_file)
    processor.process_directory(input_pcaps, features_file)
    return features_file
````

````python
@dsl.component(
    base_image='amitinfo2k/pcap-analyzer:latest',
    #packages_to_install=['pandas', 'scikit-learn']    
)
def train_model(
    features_file: str,
    output_dir: str
):
    import os
    from pcap_analyzer.modules.model_training.trainer import PCAPModelTrainer
    from pcap_analyzer.config import load_config
    
    os.makedirs(output_dir, exist_ok=True)
    model_path = os.path.join(output_dir, 'model.joblib')
    
    config = load_config()
    trainer = PCAPModelTrainer(config)
    trainer.train(features_file, model_path)
````

````python
@dsl.pipeline(
    name='pcap-analyzer-pipeline',
    description='PCAP Analysis Pipeline'
)
def pcap_analyzer_pipeline(
    input_pcaps: str = '/data/raw_pcaps',
    test_pcaps: str = '/data/test_pcaps',
    mapping_file: str = '/data/mapping.csv',
    output_dir: str = '/outputs',
    pvc_name: str = 'pcap-storage'
):
    
    # Data Preparation
    features_dir = f"{output_dir}/features"
    prepare_task = prepare_data(
        input_pcaps=input_pcaps,
        mapping_file=mapping_file,
        output_dir=features_dir
    )

     # Mount the PVC to the step and set image pull policy
    from kfp.kubernetes import mount_pvc, set_image_pull_policy
    mount_pvc(
        task=prepare_task,
        pvc_name=pvc_name,
        mount_path='/data/'
    )
    set_image_pull_policy(prepare_task, 'IfNotPresent')

    # Model Training
    model_dir = f"{output_dir}/model"
    train_task = train_model(
        features_file=prepare_task.output,
        output_dir=model_dir
    ).after(prepare_task)
   
    set_image_pull_policy(train_task, 'IfNotPresent')
    
````

````python
# Compile the pipeline
kfp.compiler.Compiler().compile(
    pipeline_func=pcap_analyzer_pipeline,
    package_path='pcap_analyzer_pipeline.yaml'
)
````
